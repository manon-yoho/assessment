```python

import json
import requests

# Fetch data from the API
url = "https://jsonplaceholder.typicode.com/users"
data = requests.get(url).json()

# Print the first 3 users
for u in data[:3]:
    print(u["id"], u["name"], u["email"])

# Save as a JSON file
with open(
    "users.json", "w", encoding="utf-8"
) as f:
  json.dump(data, f, indent=4)
  
```

```python
url = "https://jsonplaceholder.typicode.com/users"

try:
  response = requests.get(url, timeout=5)
  response.raise_for_status()
  users_data = response.json()

  parsed_users = []
  for u in users_data:
    parsed_users.append({
        "id": u.get("id"),
        "name": u.get("name"),
        "email": u.get("email"),
        "city": u.get("address", {}).get("city"),
    })

  for u in parsed_users[:3]:
    print(f"{u['id']} - {u['name']} - {u['email']}")

  with open("users.json", "w", encoding="utf-8") as f:
    json.dump(parsed_users, f, indent=4)

  # 404 error check
  resp_404 = requests.get(f"{url}/9999", timeout=5)
  if resp_404.status_code == 404:
    print("/users/9999 -> 404 Not found")

except Timeout:
  print("Request timed out.")
except ConnectionError:
  print("Connection error.")
except requests.exceptions.RequestException as e:
  print(f"Error: {e}")
```
